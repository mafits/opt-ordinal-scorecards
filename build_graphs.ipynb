```python
def acc_vs_size_plot(dataset_name, accuracies, model_sizes):
    model_names = ['Logistic Regression', 'SVM', 'Adaptive Lasso', 'skglm']
    colors = ['red', 'green', 'blue', 'orange']

    '''
    # RGB values corresponding to the colors
    rgb_colors = [(1, 0, 0),      # red
                (0, 1, 0),      # green
                (0, 0, 1),      # blue
                (1, 0.65, 0)]   # orange

    '''

    plt.figure(figsize=(10, 6))
    plt.scatter(model_sizes, accuracies, color=colors, s=150)

    x_limits = plt.xlim()
    y_limits = plt.ylim()

    # add dashed lines from each point to the axes
    for x, y, c in zip(model_sizes, accuracies, colors):
        plt.axhline(y, xmin=0, xmax=(x - x_limits[0]) / (x_limits[1] - x_limits[0]),
                    color=c, linestyle='--', linewidth=1, alpha=1)
        plt.axvline(x, ymin=0, ymax=(y - y_limits[0]) / (y_limits[1] - y_limits[0]),
                    color=c, linestyle='--', linewidth=1, alpha=1)

    plt.title(dataset_name)
    plt.xlabel('Model Size')
    plt.ylabel('Test Accuracy')
    plt.grid()
    plt.show()
```

# comparing discretisation techniques

```python
model_names = ['CAIM', 'Infinitesimal Bins']
colors = ['blue', 'orange']

fig, ax = plt.subplots(figsize=(5, 1))

for name, color in zip(model_names, colors):
    ax.scatter([], [], color=color, s=150, label=name)

ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis('off')

ax.legend(loc='center', frameon=False, fontsize=12, ncol=4)

plt.tight_layout()
plt.show()
```

```python
def caim_vs_infbins(dataset_name, accuracies, model_sizes):
    caim_acc = accuracies[:4]
    caim_size = model_sizes[:4]
    inf_acc = accuracies[4:]
    inf_size = model_sizes[4:]

    plt.figure(figsize=(10, 6))
    plt.scatter(caim_size, caim_acc, color='blue', label='CAIM', s=100)
    plt.scatter(inf_size, inf_acc, color='orange', label='Infinitesimal Bins', s=100)
    plt.xlabel('Model Size')
    plt.ylabel('Accuracy')
    plt.title(dataset_name)
    plt.grid(True)
    plt.show()
```

# comparing encoding techniques

```python
model_names = ['One-Hot Encoding', 'Differential Coding']
colors = ['red', 'green']

fig, ax = plt.subplots(figsize=(5, 1))

for name, color in zip(model_names, colors):
    ax.scatter([], [], color=color, s=150, label=name)

ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis('off')

ax.legend(loc='center', frameon=False, fontsize=12, ncol=4)

plt.tight_layout()
plt.show()
```

```python
def onehot_vs_diffcod(dataset_name, accuracies, model_sizes):
    onehot_acc = accuracies[:4]
    onehot_size = model_sizes[:4]
    diff_acc = accuracies[4:]
    diff_size = model_sizes[4:]

    plt.figure(figsize=(10, 6))
    plt.scatter(onehot_size, onehot_acc, color='red', label='One-Hot Encoding', s=100)
    plt.scatter(diff_size, diff_acc, color='green', label='Differential Coding', s=100)
    plt.xlabel('Model Size')
    plt.ylabel('Accuracy')
    plt.title(dataset_name)
    plt.grid(True)
    plt.show()
```
